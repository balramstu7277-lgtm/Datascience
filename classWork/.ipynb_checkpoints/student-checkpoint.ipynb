```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("student-por.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python

print(df.describe())
```

```python
df.plot(x="Name", y=["Math", "Science"], kind="bar")
plt.title("Student Marks Comparison")
plt.show()
```

```python
df.columns
```

**Grade Distribution**

```python
sns.histplot(df["G3"], bins=20)
plt.title("Final Grade Distribution")
plt.show()
```

**Final Marks Distribution**

```python
sns.histplot(df["G3"], bins=20, kde=True)
plt.title("Final Grade Distribution")
plt.show()
```

**Study Time vs Final Marks**

```python
sns.scatterplot(x="studytime", y="G3", data=df)
plt.title("Study Time vs Final Grade")
plt.show()
```

**Gender vs Marks**

```python
sns.boxplot(x="sex", y="G3", data=df)
plt.title("Gender vs Final Grade")
plt.show()
```

**Failures vs Marks**

```python
sns.barplot(x="failures", y="G3", data=df)
plt.title("Failures vs Final Grade")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```
