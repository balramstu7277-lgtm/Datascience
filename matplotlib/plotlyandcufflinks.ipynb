```python
import cufflinks as cf
import plotly.offline as pyo

pyo.init_notebook_mode(connected=True)
cf.go_offline()
```

```python
import seaborn as sns
import plotly.express as px

df = sns.load_dataset('tips')

px.bar(df, x='day', y='tip')
```

```python
px.line(df, x='day', y='tip')
```

```python
import plotly.express as px

fig = px.scatter(df, 
                 x='total_bill', 
                 y='tip', 
                 color='day',      # different colors
                 size='size')      # bubble size

fig.show()
```

```python
import plotly.express as px

fig = px.box(df, 
                 x='total_bill', 
                 y='day', 
                 color='day',      # different colors
                 )     
fig.show()
```
