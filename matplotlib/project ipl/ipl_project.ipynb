 # sum Basic Question

1. which team won the most matches

```python
match_winner= df['match_winner'].value_counts()
sns.barplot(y=match_winner.index,x = match_winner.values,palette='viridis')
plt.title("Most Match win by team")
```

2. Toss decision Trends

```python
sns.countplot(x = df['toss_decision'], palette= 'rainbow')
plt.title("Toss Decision")
```

3 Toss winner vs Match winner 

```python
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
persentage = count*100/df.shape[0]
persentage.round(2)
```

3.Hoe do team win?(Run vs wicket)

```python
sns.countplot(x=df['won_by'])
plt.title("Won By")
```

**KEY PLAYER PREFORMANCES**

1 Most "player of the match " Awards

```python
df.head()
```

```python
count = df["player_of_the_match"].value_counts().head(10)
sns.barplot(y = count.index , x = count.values,palette='mako')
plt.title("Top 10 players with man of the match")
```

2 Top Scorers

```python
high = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
high
```

```python
high.plot(kind = 'barh')
```

10 Best Bowling figures

```python
df['higest_Wicket'] = df['best_bowling_figure'].apply(lambda x:x.split('--')[0])
```

```python
df['higest_Wicket'] = df['higest_Wicket'].astype(int)
```

```python
top_bowler = df.groupby('best_bowling')['higest_Wicket'].sum().head(10)
```

```python
top_bowler.sort_values(ascending=False)
top_bowler.plot(kind='barh')
```
